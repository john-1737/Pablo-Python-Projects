```python
import matplotlib.pyplot as plot
import numpy as num
import ipywidgets as wids
from IPython.display import display
```

```python
xs = num.linspace(0, 20*num.pi, 1000)
ys = num.sin(xs) * 1
y2 = num.sin(xs) * 1

fig, ax = plot.subplots(figsize=(12, 4))
line1, = ax.plot(xs, ys)
line2, = ax.plot(xs, y2)

def update_plot(a, f, t):
    y2 = a * num.sin(f * xs - t)
    line2.set_ydata(y2)
    fig.canvas.draw_idle()
    

a_slider = wids.FloatSlider(value=0.5, min=0.5, max=3.0, description='a')
f_slider = wids.FloatSlider(value=0.5, min=0.5, max=3.0, description='f')
t_slider = wids.FloatSlider(value=0.5, min=0.5, max=60.0, description='t')
out = wids.interactive_output(update_plot, {'a': a_slider, 'f': f_slider, 't': t_slider})
ctrls = wids.VBox([a_slider, f_slider, t_slider])
display(ctrls, out)
```
