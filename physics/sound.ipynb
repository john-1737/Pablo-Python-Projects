```python
from matplotlib import pyplot as plot
import numpy as num
import wave
```

```python
xs = num.linspace(0, 2*num.pi, 200)
high = num.sin(xs)
low = num.sin(xs * 20)
fig, ax = plot.subplots()
line1, = ax.plot(xs, high, label="Low")
line2, = ax.plot(xs, low, label='High')
ax.legend()
```

```python
sRate = 44100
nSamples = sRate * 5
xs = num.arange(nSamples)/float(sRate)
vals = num.sin(2.0*num.pi*221*xs)
data = num.array(vals*32767, 'int16').tobytes()
with wave.open('sine221.wav', 'wb') as f:
    f.setparams((1, 2, sRate, nSamples, 'NONE', 'uncompressed'))
    f.writeframes(data)
print(data[:5])
```

```python
sRate = 44100
nSamples = sRate * 5
xs = num.arange(nSamples)/float(sRate)
vals = num.sin(2.0*num.pi*221*xs)
vals2 = num.sin(2.0*num.pi*220*xs)
vals3 = num.sin(2.0*num.pi*225*xs)
fig, ax = plot.subplots()
ax.plot(xs, vals + vals2 + vals3)
```
